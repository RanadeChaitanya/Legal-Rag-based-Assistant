```python
from emergency.gate import EmergencyGate
from cli.render import emergency_hit_to_response, render_response

gate = EmergencyGate()
query = "My brother was arrested by police without warrant"
hit = gate.evaluate(query)

if hit:
    response = emergency_hit_to_response(hit, disclaimer="General legal info only.")
    render_response(response)
```

```python
from server.verifier import GroundingVerifier

evidence = [{
    "chunk_id": "bnss:sec_58#0",
    "ref": "bnss:sec_58",
    "text_raw": "Person arrested to be taken before Magistrate within twenty-four hours."
}]

verifier = GroundingVerifier(evidence)

# Validates verbatim quote, existence, ref match, and figure bounds
result = verifier.verify_answer(
    sections=[{
        "heading": "Arrest Timeline",
        "content": "The person must be produced within twenty-four hours.",
        "citations": [{
            "chunk_id": "bnss:sec_58#0",
            "ref": "bnss:sec_58",
            "quote": "taken before Magistrate within twenty-four hours"
        }]
    }]
)

print("Verification Passed:", result.ok)
```
